```python
import matplotlib.pyplot as plot
import seaborn as sns
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler

dataset=pd.read_csv('/content/sample_data/customers_dataset.csv')

#before logskew

sns.histplot(data=dataset, x='Gender')
plot.show()
sns.histplot(data=dataset, x='Age', bins=10)
plot.show()
sns.histplot(data=dataset, x='Annual Income (k$)', bins=10)
plot.show()
sns.histplot(data=dataset, x='Spending Score (1-100)', bins=10)
plot.show()

#logskew

dataset['Age_skewlog'] = np.log(dataset['Age'])
dataset['Annual Income (k$)_skewlog'] = np.log(dataset['Annual Income (k$)'])
dataset['Spending Score (1-100)_skewlog'] = np.log(dataset['Spending Score (1-100)'])

#after logskew

sns.histplot(data=dataset, x='Age_skewlog', bins=10)
plot.show()
sns.histplot(data=dataset, x='Annual Income (k$)_skewlog', bins=10)
plot.show()
sns.histplot(data=dataset, x='Spending Score (1-100)_skewlog', bins=10)
plot.show()

#standardscaler

scaler = StandardScaler()
dataset[['Age_standardscaler', 'Annual Income (k$)_standardscaler', 'Spending Score (1-100)_standardscaler']] = scaler.fit_transform(dataset[['Age', 'Annual Income (k$)', 'Spending Score (1-100)']])
sns.histplot(data=dataset, x='Age_standardscaler', bins=10)
plot.show()
sns.histplot(data=dataset, x='Annual Income (k$)_standardscaler', bins=10)
plot.show()
sns.histplot(data=dataset, x='Spending Score (1-100)_standardscaler')
plot.show()

#minmaxscaler

scaler=MinMaxScaler()
dataset[['Age_minmaxscaler', 'Annual Income (k$)_minmaxscaler', 'Spending Score (1-100)_minmaxscaler']] = scaler.fit_transform(dataset[['Age', 'Annual Income (k$)', 'Spending Score (1-100)']])
sns.histplot(data=dataset, x='Age_minmaxscaler', bins=10)
plot.show()
sns.histplot(data=dataset, x='Annual Income (k$)_minmaxscaler', bins=10)
plot.show()
sns.histplot(data=dataset, x='Spending Score (1-100)_minmaxscaler')

#square root transformation

dataset['Age_sqrt'] = np.sqrt(dataset['Age'])
dataset['Annual Income (k$)_sqrt'] = np.sqrt(dataset['Annual Income (k$)'])
dataset['Spending Score (1-100)_sqrt'] = np.sqrt(dataset['Spending Score (1-100)'])

#after square root

sns.histplot(data=dataset, x='Age_sqrt', bins=10, color='#ff9896')
plot.title('Distribution of Age (Square Root Transformed)', fontsize=14)
plot.xlabel('Age (Square Root Transformed)', fontsize=12)
plot.ylabel('Count', fontsize=12)
plot.tight_layout()
plot.show()

sns.histplot(data=dataset, x='Annual Income (k$)_sqrt', bins=10, color='#98df8a')
plot.title('Distribution of Annual Income (k$) (Square Root Transformed)', fontsize=14)
plot.xlabel('Annual Income (k$) (Square Root Transformed)', fontsize=12)
plot.ylabel('Count', fontsize=12)
plot.tight_layout()
plot.show()

sns.histplot(data=dataset, x='Spending Score (1-100)_sqrt', bins=10, color='#c5b0d5')
plot.title('Distribution of Spending Score (1-100) (Square Root Transformed)', fontsize=14)
plot.xlabel('Spending Score (1-100) (Square Root Transformed)', fontsize=12)
plot.ylabel('Count', fontsize=12)
plot.tight_layout()
plot.show()

# Box-Cox transformation

from scipy import stats

# Apply Box-Cox, handle potential negative values or zeros if necessary
# For Box-Cox, data must be positive. We can add a small constant if needed,
# but assuming your data is suitable for now.
dataset['Age_boxcox'], _ = stats.boxcox(dataset['Age'])
dataset['Annual Income (k$)_boxcox'], _ = stats.boxcox(dataset['Annual Income (k$)'] + 1) # Add 1 to handle 0 if present
dataset['Spending Score (1-100)_boxcox'], _ = stats.boxcox(dataset['Spending Score (1-100)'])

# after Box-Cox

sns.histplot(data=dataset, x='Age_boxcox', bins=10, color='#c49c94')
plot.title('Distribution of Age (Box-Cox Transformed)', fontsize=14)
plot.xlabel('Age (Box-Cox Transformed)', fontsize=12)
plot.ylabel('Count', fontsize=12)
plot.tight_layout()
plot.show()

sns.histplot(data=dataset, x='Annual Income (k$)_boxcox', bins=10, color='#f7b6d2')
plot.title('Distribution of Annual Income (k$) (Box-Cox Transformed)', fontsize=14)
plot.xlabel('Annual Income (k$) (Box-Cox Transformed)', fontsize=12)
plot.ylabel('Count', fontsize=12)
plot.tight_layout()
plot.show()

sns.histplot(data=dataset, x='Spending Score (1-100)_boxcox', bins=10, color='#dbdb8d')
plot.title('Distribution of Spending Score (1-100) (Box-Cox Transformed)', fontsize=14)
plot.xlabel('Spending Score (1-100) (Box-Cox Transformed)', fontsize=12)
plot.ylabel('Count', fontsize=12)
plot.tight_layout()
plot.show()
```

```python
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plot
from mpl_toolkits.mplot3d import Axes3D
import seaborn as sns

dataset=pd.read_csv('/content/sample_data/customers_dataset.csv')

#K means clustering

X = dataset[['Age', 'Spending Score (1-100)']].values
k=5
kmeans=KMeans(n_clusters=k,random_state=42)
kmeans.fit(X)
cluster_labels=kmeans.labels_
cluster_centers=kmeans.cluster_centers_
dataset['cluster'] = kmeans.labels_
for cluster_id in range(k):
    cluster_data = dataset[dataset['cluster'] == cluster_id]
    plot.scatter(cluster_data['Age'], cluster_data['Spending Score (1-100)'], label=f'Cluster {cluster_id+1}')
plot.xlabel('Age')
plot.ylabel('Spending Score (1-100)')
plot.title('K-Means Clustering')
plot.legend()
plot.show()

cluster_centers = kmeans.cluster_centers_
print("Cluster Centers:")
print(cluster_centers)

#elbow plot

inertia = []
for k in range(1, 11):  # Test k from 1 to 10
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)

plot.plot(range(1, 11), inertia, marker='o')
plot.xlabel('Number of Clusters (k)')
plot.ylabel('Inertia')
plot.title('Elbow Method for Optimal k')
plot.show()


X = dataset[['Annual Income (k$)', 'Spending Score (1-100)']].values
k=5
kmeans=KMeans(n_clusters=k,random_state=42)
kmeans.fit(X)
cluster_labels=kmeans.labels_
cluster_centers=kmeans.cluster_centers_
dataset['cluster'] = kmeans.labels_
for cluster_id in range(k):
    cluster_data = dataset[dataset['cluster'] == cluster_id]
    plot.scatter(cluster_data['Annual Income (k$)'], cluster_data['Spending Score (1-100)'], label=f'Cluster {cluster_id+1}')
plot.xlabel('Annual Income (k$)')
plot.ylabel('Spending Score (1-100)')
plot.title('K-Means Clustering')
plot.legend()
plot.show()

cluster_centers = kmeans.cluster_centers_
print("Cluster Centers:")
print(cluster_centers)

#elbow plot

inertia = []
for k in range(1, 11):  # Test k from 1 to 10
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)

plot.plot(range(1, 11), inertia, marker='o')
plot.xlabel('Number of Clusters (k)')
plot.ylabel('Inertia')
plot.title('Elbow Method for Optimal k')
plot.show()

X_3d = dataset[['Age', 'Spending Score (1-100)', 'Annual Income (k$)']].values

k = 5
kmeans_3d = KMeans(n_clusters=k, random_state=42)
kmeans_3d.fit(X_3d)
dataset['cluster_3d'] = kmeans_3d.labels_

fig = plot.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

colors = ['b', 'g', 'r', 'c', 'm', 'y', 'k']

for cluster_id in range(k):
    cluster_data = dataset[dataset['cluster_3d'] == cluster_id]
    ax.scatter(cluster_data['Age'], cluster_data['Spending Score (1-100)'], cluster_data['Annual Income (k$)'],
                color=colors[cluster_id % len(colors)], label=f'Cluster {cluster_id+1}')

ax.set_xlabel('Age')
ax.set_ylabel('Spending Score (1-100)')
ax.set_zlabel('Annual Income (k$)')
ax.set_title('K-Means Clustering (3D)')
ax.legend()
plot.show()

sns.pairplot(dataset[['Age', 'Spending Score (1-100)', 'Annual Income (k$)','cluster_3d']], hue='cluster_3d', palette=colors[:k])
plot.suptitle('Pairplot of Features by Cluster', y=1.02)
plot.show()
```
